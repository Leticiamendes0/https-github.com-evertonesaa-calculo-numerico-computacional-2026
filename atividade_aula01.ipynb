```python
import math

math.sqrt(2)
```

### pergunta 1: Esse cálculo pode ser considerado um exemplo de cálculo direto?

Sim. Esse cálculo é considerado um cálculo direto porque o resultado é obtido
a partir de uma única expressão matemática, sem a necessidade de repetições
ou aproximações sucessivas. Basta aplicar a fórmula e o valor é calculado
imediatamente pelo computador. Sendo um cálculo mais simples e menos escalável.

```python
x = 1

for i in range(5):
    x = (x + 2/x) / 2
    print(x)
```

### pergunta 2: O valor obtido melhora a cada iteração? Esse processo pode ser classificado como cálculo direto ou iterativo?

Sim. A cada iteração, o valor de `x` se aproxima mais do valor correto,
mostrando uma melhora progressiva no resultado.

Esse processo é um cálculo iterativo, pois o resultado não é obtido de uma
única vez. Ele depende de repetições sucessivas, onde cada novo valor é
calculado com base no valor anterior, sendo mais flexível e escalável.


```python

round(3.1415926535, 2)

round(3.1415926535, 4)
```
