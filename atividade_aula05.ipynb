```python
# Método da Secante para f(x) = x^3 - x - 2

def f(x):
    return x**3 - x - 2

def secante(x0, x1, tol=1e-5, max_iter=100):
    print(f"{'n':<5}{'x_n-1':<15}{'x_n':<15}{'x_n+1':<15}{'erro':<15}")

    for i in range(max_iter):
        fx0 = f(x0)
        fx1 = f(x1)

        # Evitar divisão por zero
        if fx1 - fx0 == 0:
            print("Erro: divisão por zero.")
            return None

        x2 = x1 - fx1 * (x1 - x0) / (fx1 - fx0)
        erro = abs(x2 - x1)

        print(f"{i:<5}{x0:<15.6f}{x1:<15.6f}{x2:<15.6f}{erro:<15.6f}")

        if erro < tol:
            print("\nRaiz aproximada:", x2)
            return x2

        x0, x1 = x1, x2

    print("\nNão convergiu dentro do número máximo de iterações.")
    return None


# Valores iniciais
x0 = 1
x1 = 2

# Executar
secante(x0, x1)
```
